# Creating a Simple Time Series Animation using GEE's Python API

In this notebook, we try creating a time series analysis animation using satelite imagery. Time series analysis can be useful for understanding change in an area over time.

```python
#Import required libraries
import ee
import geemap
```

```python
ee.Authenticate()
```

```python
ee.Initialize()
```

```python
# Loading the Image Collection
dataset = ee.ImageCollection("COPERNICUS/S2_HARMONIZED").filterMetadata(
    'CLOUDY_PIXEL_PERCENTAGE', 'less_than', 10 
)
```

```python
# Getting the Bandnames
band_names = dataset.first().bandNames().getInfo()
print(band_names)
```

```python
# Specify a location of interest, start and end date
start_date = "2016-01-01"
end_date = "2024-12-31"
region = ee.Geometry.BBox(35.651, -0.747, 36.387, -0.082)
```

```python
# Creating an annual Composite
images = geemap.create_timeseries(
    dataset, start_date, end_date, region, frequency = 'year', reducer='median'
)
```

```python
#Displaying the time Series
m = geemap.Map()
vis_params = {min: 0, 'max': 4000, 'bands': ['B8', 'B4', 'B3']}
labels = [str(y) for y in range(2016, 2025)]
m.add_layer(images, vis_params, 'Sentinel-2', False)
m.add_time_slider(images, vis_params, time_interval=2, labels=labels)
m.center_object(region)
m
```

### Creating satellite timelaps animations

```python
m = geemap.Map(center=[-1.286389, 36.817223], zoom=6)
m
```

```python
# Defining our Region of Interest
roi = m.user_roi
if roi is None:
    roi = ee.Geometry.BBox(32.56, -3.86, 47.11, 2.46)
    m.add_layer(roi, {}, 'ROI')
    m.center_object(roi)
```

```python
#Creating our animation
timelapse = geemap.landsat_timelapse(roi, start_year=2000, 
                                    end_year=2024,
                                    #out_gif= 'output/landsat.gif',
                                    start_date="01-01",
                                   end_date="12-31",
                                   bands = ['NIR', 'Red', 'Green'],
                                  frames_per_second=5,
                                  title="Landsat Timelapse",
                                  progress_bar_color='green',
                                  mp4=True,
                                  loop = 0)
geemap.show_image(timelapse)
```
